## Introducción y Objetivos

En esta tarea semanal abordaremos el estudio analítico de diferentes topologías de filtros pasivos de segundo orden. A través del análisis topológico de redes, modelaremos el comportamiento dinámico de cada circuito para comprender cómo la disposición física de sus componentes determina su respuesta ante estímulos de frecuencia variable.

El desarrollo del trabajo se estructura en torno a los siguientes objetivos principales:

* **Obtención de Funciones de Transferencia:** Deducir analíticamente la expresión $H(s) = \frac{V_o(s)}{V_i(s)}$ para cada topología propuesta, operando con componentes normalizados para una frecuencia de resonancia $\omega_0 = 1\,\text{rad/s}$ y un factor de calidad $Q = \frac{\sqrt{2}}{2}$ (respuesta de máxima planicidad o Butterworth).

* **Cálculo de Parámetros Característicos:** Determinar el ancho de banda (BW) operativo de cada circuito fundamentado en la relación teórica entre la frecuencia central y el factor de calidad.
* **Análisis de Respuesta en Frecuencia:** Formular las expresiones analíticas de módulo y fase en régimen permanente senoidal ($s = j\omega$) para construir los Diagramas de Bode, contrastando las curvas reales con sus aproximaciones asintóticas.

## Impedancias

$$X_C = \frac{1}{j\omega C} \qquad ; \qquad X_L = j\omega L  \qquad; \qquad Z_1 = X_{L}+ R = sL + R $$

$$s = j\omega$$


## Función de transferencia

 Por divisor de tensión obtenemos la siguente expresión:

$$V_{out} = V_{in}\,\frac{\dfrac{1}{sC}}{\dfrac{1}{sC} + (sL + R)}$$

$$H(s) = \frac{V_{out}(s)}{V_{in}(s)} = \frac{\dfrac{1}{sC}}{\dfrac{1}{sC} + sL + R}$$

Multiplicando numerador y denominador por $s$:

$$H(s) = \frac{\dfrac{1}{C}}{\dfrac{1}{C} + s^2 L + sR} = \frac{\dfrac{1}{LC}}{s^2 + s\dfrac{R}{L} + \dfrac{1}{LC}}$$

Comparando con la forma estándar de un sistema de segundo orden:

$$\omega_0^2 = \frac{1}{LC} \qquad ; \qquad \frac{\omega_0}{Q} = \frac{R}{L}$$

## Valores de diseño

$$\omega_0 = 1\ \frac{\text{rad}}{\text{seg}} \qquad Q = \frac{\sqrt{2}}{2} \qquad R = 1\ \Omega$$

De $\dfrac{\omega_0}{Q} = \dfrac{R}{L}$ y $\omega_0^2 = \dfrac{1}{LC}$:

$$\boxed{L = \frac{R\,Q}{\omega_0} = \frac{\sqrt{2}}{2}\ \text{Hy}}$$

$$\boxed{C = \frac{1}{\omega_0^2 L} = \frac{2}{\sqrt{2}}\ \text{F}}$$]

## Módulo y fase

Escribiendo $H(s) = \dfrac{P}{Q}$ (con $P$ el numerador y $Q$ el denominador evaluados en $s = j\omega$):

$$|H| = \frac{|P|}{|Q|} \qquad ; \qquad \angle H = \angle P - \angle Q$$

Para un número complejo cualquiera:

$$|P|^2 = \text{Re}^2(P) + \text{Im}^2(P) \qquad ; \qquad \angle P = \tan^{-1}\left(\frac{\text{Im}(P)}{\text{Re}(P)}\right)$$

### Numerador

$$P = \frac{1}{LC} \;\Rightarrow\; \text{Re}^2(P) = \left(\frac{1}{LC}\right)^2 \;\Rightarrow\; |P| = \frac{1}{LC} \qquad ; \qquad \angle P = 0$$

### Denominador

Con $s = j\omega$:

$$Q = \left(-\omega^2 + \frac{1}{LC}\right) + j\,\omega\frac{R}{L}$$

$$|P| = \frac{1}{LC} \qquad\qquad |Q| = \sqrt{\left(-\omega^2 + \frac{1}{LC}\right)^2 + \left(\omega\frac{R}{L}\right)^2}$$

### Resultado (con $\omega_0 = 1$, $Q = \sqrt2/2$, $R = 1\,\Omega$)

Con estos valores $\dfrac{1}{LC} = 1$ y $\dfrac{R}{L} = \sqrt{2}$, entonces:

$$\boxed{|H(\omega)| = \frac{1}{\sqrt{\left(-\omega^2 + 1\right)^2 + 2\omega^2}}}$$

$$\boxed{\angle H(\omega) = -\tan^{-1}\left(\frac{\omega\,\dfrac{R}{L}}{-\omega^2 + 1}\right) = -\tan^{-1}\left(\frac{\sqrt{2}\,\omega}{1 - \omega^2}\right)}$$

# | Análisis y Resolución de Red RLC: Filtro Pasa-Altos (Circuito B)

En la presente resolución, adjunta en el PDF, se analiza el comportamiento en frecuencia de un circuito RLC pasivo. La topología consta de una resistencia y un capacitor en paralelo (rama serie), y un inductor en derivación a masa. La función de transferencia relaciona la tensión de salida $V_o(s)$ (que cae sobre el inductor) con la tensión de entrada $V_i(s)$.


<img src="data/CircuitoBjpg.png" align="center">

 Circuito equivalente con impedancia $Z_1$:
 
<img src="data/CircuitoBZ1.png" align="center">


$Z_1 = R \parallel \frac{1}{sC} = \frac{R}{sCR+1}$

$\frac{1}{Z_1} = \frac{1}{R} + sC = \frac{1+sCR}{R}$

$Z_T = Z_1 + sL = \frac{R}{sCR+1} + sL = \frac{R+s^2LCR+sL}{sCR+1}$

Aplico un divisor de tensión para hallar la función de transferencia:

$V_o(s) = V_i(s) \cdot \frac{sL}{Z_T}$

$H(s) = \frac{V_o(s)}{V_i(s)} = \frac{sL}{Z_T} = sL \cdot \frac{sCR+1}{s^2LCR+sL+R} = \frac{s^2LCR+sL}{s^2LCR+sL+R}$

$H(s) = \frac{sL(sRC+1)}{s^2LCR+sL+R}$

Divido todo por $RLC$:

$$H(s) = \frac{s^2+s\frac{1}{RC}}{s^2+s\frac{1}{RC}+\frac{1}{LC}} = \frac{s^2+s\frac{\omega_0}{Q}}{s^2+s\frac{\omega_0}{Q}+\omega_0^2}$$

$\omega_0^2 = \frac{1}{LC} \implies LC = 1 = \frac{1}{\omega_0^2} \implies L = \frac{1}{C} = \frac{1}{Q}$

$\frac{\omega_0}{Q} = \frac{1}{RC} \implies RC = \frac{Q}{\omega_0} \implies C = \frac{Q}{R\omega_0} \implies C = Q$

$$[Q = \frac{\sqrt{2}}{2} \implies C = \frac{\sqrt{2}}{2} \text{ F} \quad L = \sqrt{2} \text{ H}]$$

Entonces:

$$H(s) = \frac{s^2+\sqrt{2}s}{s^2+\sqrt{2}s+1} = \frac{s(s+\sqrt{2})}{s^2+\sqrt{2}s+1}$$

$$H(j\omega) = \frac{(j\omega)^2+\sqrt{2}j\omega}{(j\omega)^2+\sqrt{2}j\omega+1}$$

$j\omega^2 = -\omega^2$

$$H(j\omega) = \frac{-\omega^2+j\sqrt{2}\omega}{-\omega^2+j\sqrt{2}\omega+1} = \frac{-\omega^2+j\sqrt{2}\omega}{1-\omega^2+j\sqrt{2}\omega}$$

$$\vert{}H(j\omega)\vert{} = \frac{\vert{}-\omega^2+j\sqrt{2}\omega\vert{}}{\vert{}1-\omega^2+j\sqrt{2}\omega\vert{}}$$

Respuesta de Módulo:

$$\vert{}H(j\omega)\vert{} = \frac{\sqrt{(-\omega^2)^2+(\sqrt{2}\omega)^2}}{\sqrt{(1-\omega^2)^2+(\sqrt{2}\omega)^2}} = \frac{\sqrt{\omega^4+2\omega^2}}{\sqrt{(1-\omega^2)^2+2\omega^2}}$$

$$\vert{}H(j\omega)\vert{} = \sqrt{\frac{\omega^4+2\omega^2}{(1-\omega^2)^2+2\omega^2}} = \sqrt{\frac{\omega^4+2\omega^2}{1-2\omega^2+\omega^4+2\omega^2}} = \sqrt{\frac{\omega^4+2\omega^2}{1+\omega^4}}$$

Cuando $\omega \to 0$:
$$\vert{}H(j0)\vert{} = \frac{0}{1} = 0 \implies \text{En dB tiende a } -\infty$$

Cuando $\omega = 1$:
$$\vert{}H(j1)\vert{} = \sqrt{\frac{1+2}{1+1}} = \sqrt{\frac{3}{2}} = 1.225$$
$$20\log_{10}(1.225) = 1.76 \text{ dB}$$

Cuando $\omega \to +\infty$:
$$|H(j\omega)| \approx \sqrt{\frac{\omega^4}{\omega^4}}$$
$$\lim_{\omega \to +\infty} |H(j\omega)| = \sqrt{\frac{\omega^4}{\omega^4}} = 1 \implies \text{En dB tiende a 0.}$$

Respuesta de Fase:
$$\measuredangle H(j\omega) = tg^{-1}\left(\frac{\text{Im}(num)}{\text{Re}(num)}\right) - tg^{-1}\left(\frac{\text{Im}(den)}{\text{Re}(den)}\right)$$

$$\measuredangle H(j\omega) = tg^{-1}\left(\frac{\omega\sqrt{2}}{-\omega^2}\right) - tg^{-1}\left(\frac{\omega\sqrt{2}}{1-\omega^2}\right)$$

### Justificación del Tipo de Filtro (Pasa-Altos)

Tanto el análisis matemático de los límites asintóticos como la inspección física de la topología del circuito demulan de forma concluyente que nos encontramos ante un **filtro pasa-altos**. Esta afirmación se fundamenta en dos argumentos principales:

1. **Justificación Matemática (Límites de la Función de Transferencia):**
   Al evaluar el módulo de la respuesta en frecuencia en los extremos del espectro, se demostró que:
   * $\lim_{\omega \to 0} |H(j\omega)| = 0$: El circuito presenta una atenuación infinita en corriente continua y frecuencias muy bajas, es decir, no permite el paso de la señal.
   * $\lim_{\omega \to \infty} |H(j\omega)| = 1$: A medida que la frecuencia tiende a infinito, la ganancia del circuito tiende a la unidad ($0 \text{ dB}$). Esto indica que las señales de alta frecuencia pasan hacia la salida sin sufrir atenuación.

2. **Justificación Física (Topología de Componentes):**
   Observando el diagrama del Circuito B, la señal de entrada debe atravesar una rama serie compuesta por un capacitor ($C$) y un resistor ($R$) en paralelo, mientras que la salida está en paralelo con un inductor ($L$) conectado a masa. 
   * **En bajas frecuencias:** La reactancia del capacitor ($X_c = \frac{1}{\omega C}$) tiende a infinito, actuando como un circuito abierto que bloquea el paso de la señal. Simultáneamente, la reactancia del inductor ($X_L = \omega L$) tiende a cero, comportándose como un cortocircuito que deriva cualquier mínima corriente restante hacia masa, asegurando que $V_o \approx 0$.
   * **En altas frecuencias:** El comportamiento se invierte. El capacitor actúa como un cortocircuito ($X_c \to 0$), dejando pasar la señal libremente, mientras que el inductor presenta una impedancia infinita ($X_L \to \infty$), evitando que la señal se fugue a masa y permitiendo que toda la tensión caiga en la salida ($V_o \approx V_i$).
### Ancho de banda
El ancho de banda de este filtro pasa-altos se define a partir de la frecuencia de corte ($\omega_c$), donde la ganancia cae a $\frac{1}{\sqrt{2}}$ (-3 dB).


Busco $\omega_c$ donde $|H(j\omega_c)| = \frac{1}{\sqrt{2}}$

$$|H(j\omega_c)| = \sqrt{\frac{\omega_c^4+2\omega_c^2}{1+\omega_c^4}} = \frac{1}{\sqrt{2}}$$

$$\frac{\omega_c^4+2\omega_c^2}{1+\omega_c^4} = \frac{1}{2}$$

$$2\omega_c^4+4\omega_c^2 = 1+\omega_c^4$$

$$\omega_c^4+4\omega_c^2-1=0$$

$$(\omega_c^2)^2+4\omega_c^2-1=0$$

Ecuación bicuadrática, resuelvo con $\omega_c^2$ y tomo raíz positiva

$$\omega_c^2 = \frac{-4+\sqrt{16-4\cdot1\cdot(-1)}}{2} = -2+\sqrt{5}$$

$$\omega_c = \sqrt{\sqrt{5}-2}$$

$$\omega_c = 0.486 \text{ rad/s}$$

$$BW = [0.486, +\infty) \text{ rad/s}$$

$$f_c = \frac{0.486}{2\pi} = 0.077 \text{ Hz} = 77 \text{ mHz}$$

$$BW = [77 \text{ mHz}, +\infty)$$

Aunque se trate de una topología con dos elementos reactivos (circuito de segundo orden), la pendiente de subida en la banda de atenuación no es de **+40 dB/década**. Esto se debe a la distribución de sus ceros.

1. El denominador tiende a 1, ya que los términos con $s$ se vuelven despreciables.
2. En el numerador, el término $(s + \sqrt{2})$ se aproxima a la constante $\sqrt{2}$.
3. La función se reduce a $H(s) \approx s\sqrt{2}$.

Al quedar un único polo $s$ multiplicando en el numerador, la respuesta en bajas frecuencias está completamente dominada por un solo cero ubicado en el origen. Un único cero activo en esta región frecuencial produce un comportamiento de primer orden puro, determinando de forma exacta la pendiente de **+20 dB/década**.

Al observar la gráfica una década por encima de la frecuencia de corte, es decir, alrededor de $f \approx 1.59\ \text{Hz}$, la curva sólida roja de magnitud desciende por debajo de **−40 dB**.

### Justificación de la pendiente de −40 dB/década

Para una topología RLC pasa bajos, la función de transferencia canónica de segundo orden es:


$H(s) = \frac{\omega_0^2}{s^2 + 2\zeta\omega_0 s + \omega_0^2}$


Para frecuencias muy altas $ s \to \infty\ $:

1. En el denominador, el término cuadrático $s^2 $ crece mucho más rápido que los términos lineales y constantes, por lo que se vuelve dominante.
2. El numerador permanece constante e igual a $ \omega_0^2\ $.
3. Por lo tanto, la función de transferencia se aproxima a:

   $
   H(s) \approx \frac{\omega_0^2}{s^2}
   $

Como el denominador queda dominado por un término $s^2\ $, el circuito presenta el comportamiento de **dos polos** en altas frecuencias. Cada polo aporta una atenuación de aproximadamente −20 dB/década; en conjunto, la pendiente asintótica es de **−40 dB/década**.

- **Frecuencia de corte:** a $159.81\ \text{mHz}$, la ventana del cursor registra un desfase de **−90.33°**. Este valor coincide aproximadamente con el comportamiento teórico de un sistema de segundo orden, cuya fase cruza por −90° en la frecuencia natural $\omega\ $.

# | Ejercicio Bonus: Síntesis por Levantamiento de Masas

### Introducción y Marco Teórico
En esta sección se aborda el diseño del circuito utilizando una técnica de síntesis de redes conocida como **"levantamiento de masas"** (o división de admitancias). El objetivo principal de este método es otorgar libertad geométrica para reubicar los *ceros de transmisión* de un filtro, permitiendo sintetizar respuestas complejas (como un filtro biquadrático o Notch) sin alterar los *polos* originales que definen la estabilidad, la frecuencia de resonancia ($\omega_0$) y el factor de calidad ($Q$) del sistema.

**Teoría del Método:**
El principio fundamental se basa en la propiedad de las admitancias en paralelo. Si consideramos una admitancia total $Y$ conectada entre un nodo y masa, el método propone "levantar" una parte de ese componente hacia la rama serie que conecta la entrada con la salida, fraccionándolo en dos porciones mediante un coeficiente (por ejemplo, $a$, donde $0 \le a \le 1$):
*   **Rama Serie ($Y_1$):** Se asigna una fracción de la admitancia original, equivalente a $Y \cdot a$.
*   **Rama a Masa ($Y_2$):** Se deja el remanente de la admitancia en derivación, equivalente a $Y \cdot (1 - a)$.

La genialidad del método radica en que, vista desde el nodo de salida, la admitancia total sigue siendo exactamente la misma: $Y_T = Y_1 + Y_2 = Y \cdot a + Y(1-a) = Y$. Al mantener la admitancia total inalterada, la ecuación característica (el denominador del sistema) permanece idéntica a la del circuito resonante original.

**Arreglo del Circuito Propuesto:**
Aplicando esta teoría al circuito de la consigna, se parte de un tanque RLC paralelo ideal con capacitancia total $C$ e inductancia total $L$.
1.  **Fracción Capacitiva ($a$):** Se ubica un capacitor $C \cdot a$ en la rama serie y un capacitor $C(1-a)$ conectado a masa.
2.  **Fracción Inductiva ($b$):** Se ubica un inductor de valor $L/b$ en la rama serie (ya que la admitancia inductiva es inversamente proporcional a la inductancia) y un inductor de valor $\frac{L}{1-b}$ conectado a masa.
3.  **Resistencia ($R$):** Permanece íntegramente en derivación a masa, fijando el factor de amortiguamiento.

Al aplicar un divisor de tensión ($H(s) = Y_1 / Y_T$) sobre este arreglo, se obtendrá una función de transferencia paramétrica donde los valores de $a$ y $b$ dictarán la forma del numerador, permitiendo transicionar fluidamente entre un comportamiento pasa-bajos, pasa-altos o rechaza-banda.

**ESQUEMA:**

![Esquema Bonus 1](data/bc1.jpeg)

Planteo las admitancias:

$Y_1 = \frac{b}{sL} \parallel sC\cdot a = sC\cdot a + \frac{b}{sL}$

$Y_2 = sC(1-a) \parallel \frac{1-b}{sL} \parallel \frac{1}{R} = sC(1-a) + \frac{1-b}{sL} + \frac{1}{R}$

**CIRCUITO EQUIVALENTE:**

![Esquema Circuito Equivalente](data/bc2.jpeg)
Aplico un divisor de tension para hallar la función de transferencia:

$$H(s) = \frac{V_o(s)}{V_i(s)} = \frac{Y_1}{Y_1 + Y_2} = \frac{Y_1}{Y_T}$$

$$Y_T = Y_1 + Y_2 = sC\cdot a + \frac{b}{sL} + sC(1-a) + \frac{1-b}{sL} + \frac{1}{R}$$

$$Y_T = sC (a + 1 - a) + \frac{b + 1 - b}{sL} + \frac{1}{R}$$

$$Y_T = sC + \frac{1}{sL} + \frac{1}{R}$$

$\implies H(s) = \frac{Y_1}{Y_T} = \frac{sC\cdot a + \frac{b}{sL}}{sC + \frac{1}{sL} + \frac{1}{R}}$

Multiplico numerador y denominador por $sL$:

$$H(s) = \frac{s^2 C L a + b}{s^2 C L + 1 + s\frac{L}{R}}$$

Divido todo por $LC$:

$$H(s) = \frac{s^2 a + \frac{b}{LC}}{s^2 + s\frac{1}{RC} + \frac{1}{LC}} = \frac{a s^2 + b \omega_0^2}{s^2 + s\frac{\omega_0}{Q} + \omega_0^2}$$

Datos:
$\omega_0 = 1$, $R = 1\,\Omega$ y $Q = \frac{\sqrt{2}}{2}$

$\omega_0^2 = \frac{1}{LC} \implies 1^2 = \frac{1}{LC} \implies LC = 1 \implies L = \frac{1}{C}$

$\frac{\omega_0}{Q} = \frac{1}{RC} \implies \frac{1}{\sqrt{2}/2} = \frac{1}{C} \implies C = \frac{\sqrt{2}}{2} \text{ F}$

Resulta:

$$H(s) = \frac{a s^2 + b}{s^2 + s\sqrt{2} + 1}$$

**Conclusión**

El método de levantamiento de masas permitió construir un filtro biquadrático. Ajustando las fracciones $a$ y $b$ se pueden diseñar distintas funciones de transferencia:

Si $a=1$ y $b=0$, levanto todo el capacitor de la rama serie.

$\implies H(s) = \frac{s^2}{s^2 + s\sqrt{2} + 1}$ y me queda un pasa-altos puro.

Si $a=0$ y $b=1$, levanto todo el inductor de la rama serie.

$\implies H(s) = \frac{1}{s^2 + s\sqrt{2} + 1}$ y me queda un pasa-bajos puro.

Si combino valores fraccionarios de $a$ y $b$, puedo generar un filtro notch u otros comportamientos complejos.

### Análisis del Diagrama de Bode: Filtro High-Pass Notch

En la imagen obtenida de la simulación en LTspice, se observa la respuesta en frecuencia del circuito escalado a $10 \text{ kHz}$ utilizando la técnica de levantamiento de masas. Al configurar los parámetros fraccionarios en $a = 0.999$ y $b = 0.001$, el sistema adopta el comportamiento de un **filtro pasa-altos con un cero de transmisión** (High-Pass Notch).

Del análisis de la gráfica se desprenden tres regiones fundamentales que validan el modelo matemático $H(s) = \frac{a \cdot s^2 + b\omega_0^2}{s^2 + s\frac{\omega_0}{Q} + \omega_0^2}$:

1.  **Comportamiento en Altas Frecuencias (Banda de Paso):**
    En el extremo derecho de la gráfica, para frecuencias muy por encima de la resonancia, se observa que la curva de magnitud (verde continuo) se estabiliza de forma asintótica en **$0 \text{ dB}$**. Esto responde al límite de la función de transferencia cuando $s \to \infty$, donde la ganancia tiende al valor del parámetro $a$. Al ser $a = 0.999$, la magnitud es prácticamente la unidad ($0 \text{ dB}$), permitiendo el paso irrestricto de las altas frecuencias.

2.  **Comportamiento en Bajas Frecuencias (Banda de Atenuación):**
    En el extremo izquierdo de la gráfica, para frecuencias muy bajas ($s \to 0$), la curva de magnitud inicia completamente plana en el nivel de **$-60 \text{ dB}$**. Esto concuerda perfectamente con el límite teórico donde la ganancia tiende al valor del parámetro $b$. En escala logarítmica, $20\log_{10}(0.001) = -60 \text{ dB}$.

3.  **El "Pozo" o Cero de Transmisión (Notch):**
    El rasgo más destacado de la gráfica es la profunda muesca de atenuación que se genera en la banda de transición. Este "pozo" ocurre matemáticamente cuando el numerador de la función de transferencia se anula, lo que dicta un cero de transmisión en $f_z = f_0 \sqrt{\frac{b}{a}}$.
    Al evaluar con los parámetros de diseño ($f_0 = 10 \text{ kHz}$, $a = 0.999$, $b = 0.001$), la frecuencia teórica del cero es $f_z \approx 316.38 \text{ Hz}$. 
    
    El cursor de la simulación corrobora este fenómeno con extrema precisión: al posicionarse en $f \approx 317.48 \text{ Hz}$, se registra una caída abrupta de la magnitud hasta **$-104.1 \text{ dB}$**. Simultáneamente, en esta misma frecuencia se produce una rotación violenta en la curva de fase (línea punteada), la cual salta abruptamente para ubicarse en **$177.39^\circ$** (prácticamente $180^\circ$), lo cual es el comportamiento característico e inequívoco al atravesar un cero sobre el eje imaginario ($j\omega$).

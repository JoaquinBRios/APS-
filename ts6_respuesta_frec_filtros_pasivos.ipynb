## 1. Introducción y Objetivos

En esta tarea semanal abordaremos el estudio analítico de diferentes topologías de filtros pasivos de segundo orden. A través del análisis topológico de redes, modelaremos el comportamiento dinámico de cada circuito para comprender cómo la disposición física de sus componentes determina su respuesta ante estímulos de frecuencia variable.

El desarrollo del trabajo se estructura en torno a los siguientes objetivos principales:

* **Obtención de Funciones de Transferencia:** Deducir analíticamente la expresión $H(s) = \frac{V_o(s)}{V_i(s)}$ para cada topología propuesta, operando con componentes normalizados para una frecuencia de resonancia $\omega_0 = 1\,\text{rad/s}$ y un factor de calidad $Q = \frac{\sqrt{2}}{2}$ (respuesta de máxima planicidad o Butterworth).
* **Clasificación del Sistema:** Identificar el tipo de filtro (Pasa-bajos, Pasa-altos, Pasa-banda) mediante la evaluación de sus polos y ceros en el plano complejo, complementando la matemática con el análisis del comportamiento asintótico de las impedancias reactivas ($s \to 0$ y $s \to \infty$).
* **Cálculo de Parámetros Característicos:** Determinar el ancho de banda (BW) operativo de cada circuito fundamentado en la relación teórica entre la frecuencia central y el factor de calidad.
* **Análisis de Respuesta en Frecuencia:** Formular las expresiones analíticas de módulo y fase en régimen permanente senoidal ($s = j\omega$) para construir los Diagramas de Bode, contrastando las curvas reales con sus aproximaciones asintóticas.

**Desarrollo Analítico y Función de Transferencia**
Aplicando el divisor de tensión con las impedancias correspondientes, se obtuvo la siguiente función de transferencia:

$$ H(s) = \frac{s^2 + s\frac{1}{RC}}{s^2 + s\frac{1}{RC} + \frac{1}{LC}} $$

Se identifican los parámetros característicos del sistema de segundo orden:
*   $\omega_0^2 = \frac{1}{LC}$
*   $\frac{\omega_0}{Q} = \frac{1}{RC}$

Sustituyendo los valores de diseño requeridos ($\omega_0 = 1$ rad/s y $Q = \sqrt{2}/2$), la función de transferencia adopta su forma numérica final:

$$ H(s) = \frac{s^2 + s\sqrt{2}}{s^2 + s\sqrt{2} + 1} $$

**Evaluación de Módulo y Fase**
Evaluando en el régimen senoidal permanente ($s = j\omega$), se determinaron las expresiones para la respuesta de magnitud y fase:

*   **Módulo:** $|H(j\omega)| = \sqrt{\frac{\omega^4 + 2\omega^2}{1 + \omega^4}}$
*   **Fase:** $\phi(\omega) = \arctan\left(\frac{\omega\sqrt{2}}{-\omega^2}\right) - \arctan\left(\frac{\omega\sqrt{2}}{1 - \omega^2}\right)$

Como se muestra en el archivo PDF adjunto (`ts6_b2.pdf`), el análisis gráfico de polos y ceros confirma el comportamiento de la fase.

*   **Polos y Ceros:** Se identifican dos ceros, uno en el origen ($z_1 = 0$) y otro real en $z_2 = -\sqrt{2}$. Los polos son complejos conjugados ubicados en $p_{1,2} = -0.707 \pm j0.707$.
*   **Comportamiento de la Fase:**
    *   **Bajas Frecuencias ($\omega \to 0^+$):** La fase está dominada por el cero en el origen, comenzando en $\pi/2$ ($+90^\circ$).
    *   **Frecuencia de Resonancia ($\omega = \omega_0 = 1$):** El cero real aporta fase, mientras que el polo complejo ($p_1$) provoca una rotación abrupta que "tira" la fase hacia abajo, generando el cambio brusco.
    *   **Altas Frecuencias ($\omega \to \infty$):** Las contribuciones de los ceros ($\pi$) y los polos ($-\pi$) se cancelan, estabilizando la fase en 0.

El ancho de banda de este filtro pasa-altos se define a partir de la frecuencia de corte ($\omega_c$), donde la ganancia cae a $\frac{1}{\sqrt{2}}$ (-3 dB).

Igualando la magnitud y resolviendo la ecuación bicuadrática ($\omega_c^4 + 4\omega_c^2 - 1 = 0$), se obtiene:
*   $\omega_c = \sqrt{\sqrt{5} - 2} \approx 0.486 \text{ rad/s}$

Expresado en frecuencia (Hz):
*   $f_c = \frac{0.486}{2\pi} \approx 77 \text{ mHz}$

Por lo tanto, el Ancho de Banda es:
*   **$BW = [0.486, \infty) \text{ rad/s}$**
*   **$BW = [77 \text{ mHz}, \infty)$**

A partir del análisis en frecuencia, se verificaron los siguientes puntos clave:

*   **Sobrepico de Resonancia:** El cursor posicionado en la frecuencia de resonancia teórica $\omega_0 = 1\text{ rad/s}$ (**159.38 mHz**) registra un sobrepico de **1.76 dB**. Esto confirma el aporte de ganancia del cero real desplazado en la zona de transición.
*   **Frecuencia de Corte:** Se verificó el punto de media potencia con una magnitud de **-3.04 dB** en la frecuencia $f_c \approx$ **77.13 mHz**.
*   **Pendiente de Atenuación Asintótica:** Al medir exactamente una década por debajo de la frecuencia de corte (en $f \approx$ **7.74 mHz**), la magnitud desciende a **-23.24 dB**. La variación arrojada por el simulador entre estos dos cursores es de **-20.2 dB**. Esto demuestra empíricamente una caída de **20 dB** a lo largo de exactamente una década de frecuencia.

**Justificación de la pendiente de +20 dB/década:**
Aunque se trate de una topología con dos elementos reactivos (circuito de segundo orden), la pendiente de subida en la banda de atenuación no es de **+40 dB/década**. Esto se debe a la distribución de sus ceros. 
Evaluando la función de transferencia $H(s) = \frac{s(s + \sqrt{2})}{s^2 + s\sqrt{2} + 1}$ para frecuencias muy bajas ($s \to 0$):
1. El denominador tiende a 1, ya que los términos con $s$ se vuelven despreciables.
2. En el numerador, el término $(s + \sqrt{2})$ se aproxima a la constante $\sqrt{2}$.
3. La función se reduce a $H(s) \approx s\sqrt{2}$.

Al quedar un único polo $s$ multiplicando en el numerador, la respuesta en bajas frecuencias está completamente dominada por un solo cero ubicado en el origen. Un único cero activo en esta región frecuencial produce un comportamiento de primer orden puro, determinando de forma exacta la pendiente de **+20 dB/década**.

**Comportamiento de la Fase:**
La curva de fase en la simulación respalda el análisis geométrico de polos y ceros:
*   En las bajas frecuencias ($f \approx 7.74 \text{ mHz}$), la fase registra **87.19°**, demostrando que la curva parte asintóticamente desde los **+90°** dictados por el cero en el origen.
*   En la zona de paso (altas frecuencias), se observa que la fase tiende a asintóticamente a **0°**, confirmando la cancelación angular total entre los polos y los ceros para valores grandes de $\omega$.
